```python
def suggest_rf_params(trial):
    params = {
        # Try different numbers of decision trees between 100 and 500
        "n_estimators": trial.suggest_int("n_estimators", 100, 500, step=50),
        # Try tree depths from 3 (shallow/simple) up to 16 (deep/complex)
        "max_depth": trial.suggest_int("max_depth", 3, 16),
        # Minimum number of samples required to split an internal decision node
        "min_samples_split": trial.suggest_int("min_samples_split", 2, 16),
        # Minimum number of samples required to be at a leaf branch ending
        "min_samples_leaf": trial.suggest_int("min_samples_leaf", 1, 8),
        # Try using different percentages of features (40% to 100%) for each tree
        "max_features": trial.suggest_float("max_features", 0.4, 1.0),
    }
    return params
```

```python
rf_results = run_tree_configurations("Random Forest", suggest_rf_params, prepared, RUN_DIR / "rf_validation.joblib")
```
